```python
def fix_new_token(sample):
    if '''_{"code":200,"status":true,"model":"gpt-3.5-turbo","gpt":"\'''' in sample['en']:
        sample['en'] = sample['en'].replace('''_{"code":200,"status":true,"model":"gpt-3.5-turbo","gpt":"\'''','')
    if ''',"original":null}''' in sample['en']:
        sample['en'] = sample['en'].replace(''',"original":null}''','')
    
    sample['en'] = sample['en'].replace('\\n','\n')
    sample['en'] = sample['en'].replace('\'','')
    sample['vi'] = sample['vi'].replace('\n','<n>')
    sample['en'] = sample['en'].replace('\n','<n>')
    return sample

raw_datasets = raw_datasets.map(fix_new_token)
```

```python
raw_datasets = raw_datasets.train_test_split(test_size=0.05)
```

```python
import datasets
import random
import pandas as pd
from IPython.display import display, HTML

def show_random_elements(dataset, num_examples=5):
    assert num_examples <= len(dataset), "Can't pick more elements than there are in the dataset."
    picks = []
    for _ in range(num_examples):
        pick = random.randint(0, len(dataset)-1)
        while pick in picks:
            pick = random.randint(0, len(dataset)-1)
        picks.append(pick)
    
    df = pd.DataFrame(dataset[picks])
    for column, typ in dataset.features.items():
        if isinstance(typ, datasets.ClassLabel):
            df[column] = df[column].transform(lambda i: typ.names[i])
    display(HTML(df.to_html()))
```

# Processing data

```python
from transformers import AutoTokenizer
    
tokenizer = AutoTokenizer.from_pretrained('VietAI/envit5-translation',skip_special_tokens=False)
```

```python
tokenizer.truncation_side='left'
```

```python
prefix = ""
```

```python
max_input_length = 1024
max_target_length = 1024

def preprocess_function(examples):
    #inputs = ['vi: ' + doc for doc in examples["vi"]] + ['en: ' + doc for doc in examples["en"]]
    inputs = ['vi: ' + doc for doc in examples["vi"]]
    model_inputs = tokenizer(inputs, max_length=max_input_length, truncation=True)

    # Setup the tokenizer for targets
    # labels = tokenizer(text_target=['en: ' + doc for doc in examples["en"]] + ['vi: ' + doc for doc in examples["vi"]], max_length=max_target_length, truncation=True)
    labels = tokenizer(text_target=['en: ' + doc for doc in examples["en"]], max_length=max_target_length, truncation=True)

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

def preprocess_function1(examples):
    #inputs = ['vi: ' + doc for doc in examples["vi"]] + ['en: ' + doc for doc in examples["en"]]
    inputs = ['en: ' + doc for doc in examples["en"]]
    model_inputs = tokenizer(inputs, max_length=max_input_length, truncation=True)

    # Setup the tokenizer for targets
    # labels = tokenizer(text_target=['en: ' + doc for doc in examples["en"]] + ['vi: ' + doc for doc in examples["vi"]], max_length=max_target_length, truncation=True)
    labels = tokenizer(text_target=['vi: ' + doc for doc in examples["vi"]], max_length=max_target_length, truncation=True)

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs
```

```python
tokenized_datasets = raw_datasets.map(preprocess_function, batched=True)
```

```python
tokenized_datasets1 = raw_datasets.map(preprocess_function1, batched=True)
```

```python
from datasets import concatenate_datasets, load_dataset

tokenized_datasets['train'] = concatenate_datasets([tokenized_datasets['train'], tokenized_datasets1['train']])
tokenized_datasets['test'] = concatenate_datasets([tokenized_datasets['test'], tokenized_datasets1['test']])
```

```python
tokenized_datasets = tokenized_datasets.shuffle()
```

# Model finetune

```python
from transformers import AutoModelForSeq2SeqLM, DataCollatorForSeq2Seq, Seq2SeqTrainingArguments, Seq2SeqTrainer

model = AutoModelForSeq2SeqLM.from_pretrained('VietAI/envit5-translation')
```

```python
batch_size = 16
model_name = model_checkpoint.split("/")[-1]
args = Seq2SeqTrainingArguments(
    output_dir="/raid/phundh/trans_model_vi_en",
    per_device_train_batch_size=batch_size,
    gradient_accumulation_steps=1,
    learning_rate=2e-5,
    warmup_steps=400,
    weight_decay=0.01,
    save_total_limit=4,
    num_train_epochs=4,
    gradient_checkpointing=True,
    gradient_checkpointing_kwargs={'use_reentrant':False},
    fp16=True,
    save_strategy='epoch', #save_steps=500,
    evaluation_strategy = "no",
#     evaluation_strategy = "steps",
#     eval_steps=500,
#     metric_for_best_model="rouge",
#     greater_is_better=True,
#     per_device_eval_batch_size=batch_size,
#     predict_with_generate=True,
#     generation_max_length=max_target_length,
#     load_best_model_at_end=True,
    push_to_hub=False,
    logging_steps=25,
#     metric_for_best_model="wer",
#     greater_is_better=False,
#     ignore_data_skip=True
)
```

```python
data_collator = DataCollatorForSeq2Seq(tokenizer, model=model)
```

```python
import nltk
import numpy as np

def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    decoded_preds = tokenizer.batch_decode(predictions, skip_special_tokens=True)
    # Replace -100 in the labels as we can't decode them.
    labels = np.where(labels != -100, labels, tokenizer.pad_token_id)
    decoded_labels = tokenizer.batch_decode(labels, skip_special_tokens=True)
    
    # Rouge expects a newline after each sentence
    decoded_preds = ["\n".join(nltk.sent_tokenize(pred.strip())) for pred in decoded_preds]
    decoded_labels = ["\n".join(nltk.sent_tokenize(label.strip())) for label in decoded_labels]
    
    # Note that other metrics may not have a `use_aggregator` parameter
    # and thus will return a list, computing a metric for each sentence.
    result = metric.compute(predictions=decoded_preds, references=decoded_labels, use_stemmer=True, use_aggregator=True)
    # Extract a few results
    result = {key: value * 100 for key, value in result.items()}
    
    # Add mean generated length
    prediction_lens = [np.count_nonzero(pred != tokenizer.pad_token_id) for pred in predictions]
    result["gen_len"] = np.mean(prediction_lens)
    
    return {k: round(v, 4) for k, v in result.items()}
```

```python
trainer = Seq2SeqTrainer(
    model,
    args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["test"],
    data_collator=data_collator,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics
)
```
