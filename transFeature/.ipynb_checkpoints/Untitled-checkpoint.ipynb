```python
list_dir = ['data','data_5k_10k','data_10k_15k','data_15k_20k','data_20k_25k','data_25k_30k','data_30k_35k']
```

```python
import os
import json
from tqdm import tqdm
import random

list_data = []
for dir in list_dir:
    for data_path in tqdm(os.listdir(dir)):
        if 'json' not in data_path:
            continue
        with open(f'{dir}/{data_path}', 'r') as f:
            ds = json.load(f)
            coin = random.choice([0,0,0,1,1])
            try:
                if coin == 1:
                    prev_en = ''
                    prev_vi = ''
                    for i in range(len(ds)):
                        data = ds[i]
                        if data['en'] == '': continue
                        if """kí tự xuống dòng và không sinh thêm kí tự '*', '**':""" in data['en']:
                            data['en'] = data['en'].split("""kí tự xuống dòng và không sinh thêm kí tự '*', '**':""")[1]
                        if '''được dịch sang tiếng Anh như sau:''' in data['en']:
                            data['en'] = data['en'].split('''được dịch sang tiếng Anh như sau:''')[1]
                        data['en'] = data['en'].replace('Here is the output:\n','').replace('\n\n','\n').replace('Here is the translated output:\n','')
                        if "'" == data['en'][0]:
                            data['en'] = data['en'][1:]
                        if i != len(ds) - 1:
                            prev_en += data['en'] + '\n'
                            prev_vi += data['vi'] + '\n'
                        else:
                            prev_en += data['en']
                            prev_vi += data['vi']
                        if len(prev_en.split(' ')) >= 300:
                            list_data.append({'vi' : prev_vi, 'en' : prev_en})
                            prev_vi = ''
                            prev_en = ''
                        if i == len(ds) - 1 and prev_en != '':
                            list_data.append({'vi' : prev_vi, 'en' : prev_en})
                else:
                    for i in range(len(ds)):
                        data = ds[i]
                        if """kí tự xuống dòng và không sinh thêm kí tự '*', '**':""" in data['en']:
                            data['en'] = data['en'].split("""kí tự xuống dòng và không sinh thêm kí tự '*', '**':""")[1]
                        if '''được dịch sang tiếng Anh như sau:''' in data['en']:
                            data['en'] = data['en'].split('''được dịch sang tiếng Anh như sau:''')[1]
                        data['en'] = data['en'].replace('Here is the output:\n','').replace('\n\n','\n').replace('Here is the translated output:\n','')
                        if "'" == data['en'][0]:
                            data['en'] = data['en'][1:]
                        list_data.append(data)
            except:
                print(data)
```

```python
yte = yte.remove_columns(['origin'])
```

```python
phapluat = phapluat.remove_columns(['origin'])
```

```python
from datasets import concatenate_datasets, load_dataset

data1 = concatenate_datasets([vanhoc['train'], news['train']])
data1 = concatenate_datasets([data1, yte['train']])
data1 = concatenate_datasets([data1, phapluat['train']])
```

```python
data1 = data1.filter(lambda example : len(example['en'].split(' ')) > 25)
```

```python
data2 = Dataset.from_list(list_data)
```

```python
data_final = concatenate_datasets([data2, data1])
```

```python
data_final = data_final.shuffle()
```
